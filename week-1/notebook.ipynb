```python
column_with_missing = df.isnull().any().sum()
print(f"column total with missing value: {column_with_missing}")
```

```python
max_efisiensi = df[df['origin'] == 'Asia']['fuel_efficiency_mpg'].max()
print(f"Maximum fuel efficiency from Asia: {max_efisiensi}")
```

```python
# Langkah 1: Median horsepower (sebelum fillna)
median_awal = df['horsepower'].median()
print(f"Median awal          : {median_awal}")
```

```python
# Langkah 2: Nilai paling sering (modus)
modus = df['horsepower'].mode()[0]
print(f"Modus (most frequent): {modus}")
```

```python
# Langkah 3: Fill missing values dengan modus
df['horsepower'] = df['horsepower'].fillna(modus)
```

```python
# Langkah 4: Median setelah fillna
median_akhir = df['horsepower'].median()
print(f"Median setelah fillna: {median_akhir}")

# Bandingkan
print(f"\nPerubahan: {median_akhir - median_awal}")
```

```python
# 1. Select all cars from Asia
asia = df[df['origin'] == 'Asia']

# 2. Select only columns vehicle_weight and model_year
subset = asia[['vehicle_weight', 'model_year']]

# 3. Select the first 7 values
subset = subset.head(7)

# 4. Get underlying NumPy array → X
X = subset.values
print("X shape:", X.shape)
print(X)

# 5. Matrix-matrix multiplication: X.T @ X → XTX
XTX = X.T @ X
print("\nXTX shape:", XTX.shape)

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 7. Create array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Compute w = (XTX)^-1 @ X.T @ y
w = XTX_inv @ X.T @ y
print("\nw:", w)

# 9. Sum of all elements of w
total = w.sum()
print(f"\nSum of weights: {total}")
```
